```python
import numpy as np

arr = np.random.randint(0, 51, size=20)

print("Вихідний масив:")
print(arr)

threshold = int(input("Введіть порогове значення: "))

count = np.sum(arr > threshold)

print(f"Кількість елементів більших за {threshold}: {count}")

max_value = np.max(arr)
max_position = np.argmax(arr)

print(f"Максимальне значення: {max_value}")
print(f"Позиція першої появи: {max_position}")

sorted_arr = np.sort(arr)[::-1]

print("Масив за спаданням:")
print(sorted_arr)
```

```python
import numpy as np

lower = int(input("Введіть нижню межу: "))
upper = int(input("Введіть верхню межу: "))

matrix = np.random.randint(lower, upper + 1, size=(5, 5))

print("Вихідна матриця:")
print(matrix)

diagonal = np.diag(matrix)

print("Головна діагональ:")
print(diagonal)

diagonal_sum = np.sum(diagonal)

print("Сума головної діагоналі:")
print(diagonal_sum)

modified_matrix = matrix.copy()

modified_matrix[np.triu_indices(5, k=1)] = 0

print("Матриця після обнулення елементів вище головної діагоналі:")
print(modified_matrix)
```

```python
import numpy as np

start = int(input("Введіть початок діапазону: "))
end = int(input("Введіть кінець діапазону: "))

numbers = np.arange(start, end + 1)

matrix = numbers[:30].reshape(6, 5)

print("Матриця 6x5:")
print(matrix)

row_sums = np.sum(matrix, axis=1)

print("Суми елементів у кожному рядку:")
print(row_sums)

column_max = np.max(matrix, axis=0)

print("Максимальні значення за кожним стовпчиком:")
print(column_max)
```
