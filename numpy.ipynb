```python
import numpy as np

arr = np.random.randint(0, 51, size=20)

print("Вихідний масив:")
print(arr)

threshold = int(input("Введіть порогове значення: "))

count = np.sum(arr > threshold)

print(f"Кількість елементів більших за {threshold}: {count}")

max_value = np.max(arr)
max_position = np.argmax(arr)

print(f"Максимальне значення: {max_value}")
print(f"Позиція першої появи: {max_position}")

sorted_arr = np.sort(arr)[::-1]

print("Масив за спаданням:")
print(sorted_arr)
```

```python
import numpy as np

lower = int(input("Введіть нижню межу: "))
upper = int(input("Введіть верхню межу: "))

matrix = np.random.randint(lower, upper + 1, size=(5, 5))

print("Вихідна матриця:")
print(matrix)

diagonal = np.diag(matrix)

print("Головна діагональ:")
print(diagonal)

diagonal_sum = np.sum(diagonal)

print("Сума головної діагоналі:")
print(diagonal_sum)

modified_matrix = matrix.copy()

modified_matrix[np.triu_indices(5, k=1)] = 0

print("Матриця після обнулення елементів вище головної діагоналі:")
print(modified_matrix)
```

```python
import numpy as np

lower = int(input("Введіть нижню межу: "))
upper = int(input("Введіть верхню межу: "))

arr = np.random.randint(lower, upper + 1, size=15)

print("Вихідний масив:")
print(arr)

negative_elements = arr[arr < 0]

print("Від'ємні елементи:")
print(negative_elements)

modified_arr = arr.copy()
modified_arr[modified_arr < 0] = 0

print("Масив після заміни від'ємних елементів на нулі:")
print(modified_arr)

zero_count = np.sum(modified_arr == 0)

print("Кількість нульових елементів:")
print(zero_count)
```

```python
import numpy as np

length = int(input("Введіть довжину масивів: "))

arr1 = np.random.randint(0, 11, size=length)
arr2 = np.random.randint(10, 21, size=length)

print("Перший масив:")
print(arr1)

print("Другий масив:")
print(arr2)

combined = np.concatenate((arr1, arr2))

print("Об'єднаний масив:")
print(combined)

addition = arr1 + arr2
difference = arr1 - arr2

print("Поелементне додавання:")
print(addition)

print("Поелементна різниця:")
print(difference)
```

```python
import numpy as np

rows = int(input("Введіть кількість рядків вихідної матриці: "))
cols = int(input("Введіть кількість стовпчиків вихідної матриці: "))

matrix = np.arange(1, rows * cols + 1).reshape(rows, cols)

print("Вихідна матриця:")
print(matrix)

new_rows = int(input("Введіть нову кількість рядків: "))
new_cols = int(input("Введіть нову кількість стовпчиків: "))

new_matrix = matrix.reshape(new_rows, new_cols)

print("Нова матриця:")
print(new_matrix)

row_min = np.min(new_matrix, axis=1)
row_max = np.max(new_matrix, axis=1)

print("Мінімальні значення в кожному рядку:")
print(row_min)

print("Максимальні значення в кожному рядку:")
print(row_max)

total_sum = np.sum(new_matrix)

print("Загальна сума всіх елементів:")
print(total_sum)
```
