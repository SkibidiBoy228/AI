```python
expenses = [1200, 1500, 800, 600, 1000, 400, 900]

total = sum(expenses)
average = total / len(expenses)
maximum = max(expenses)
minimum = min(expenses)

days_above_1000 = sum(1 for x in expenses if x > 1000)

print(total, average, maximum, minimum, days_above_1000)
```
