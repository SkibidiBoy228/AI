```python
math = [80, 92, 75, 60, 88]
physics = [70, 65, 90, 85, 78]
english = [95, 100, 85, 90, 87]

subjects = {
    "Math": math,
    "Physics": physics,
    "English": english
}

averages = {}

for name, scores in subjects.items():
    avg = sum(scores) / len(scores)
    averages[name] = avg

    print(
        name,
        "Average:", avg,
        "Min:", min(scores),
        "Max:", max(scores)
    )

best_subject = max(averages, key=averages.get)

print("Best subject:", best_subject)
```

|----------|--------------|-------------|--------------|

| Physics | 77.6 | 65 | 90 |
| English | 91.4 | 85 | 100 |

Найвищий середній результат: English
