```python
import matplotlib.pyplot as plt

months = ["Jan", "Feb", "Mar", "Apr", "May", "Jun"]
plan = [100, 120, 130, 140, 150, 160]
fact = [90, 125, 128, 150, 145, 170]

plt.plot(months, plan, label="План")
plt.plot(months, fact, label="Факт")

plt.xlabel("Місяці")
plt.ylabel("Продажі")
plt.title("Планові та фактичні продажі")
plt.legend()

plt.show()
```

```python
import matplotlib.pyplot as plt
import numpy as np

ages = np.random.randint(18, 70, size=100)
average_age = np.mean(ages)

plt.hist(ages, bins=10)
plt.axvline(average_age, linestyle="--", label="Середній вік")

plt.xlabel("Вік")
plt.ylabel("Кількість людей")
plt.title("Розподіл віку")
plt.legend()

plt.show()
```

```python
import matplotlib.pyplot as plt

group1 = [75, 80, 85, 90, 70, 88, 92, 78]
group2 = [60, 65, 70, 75, 80, 68, 72, 77]
group3 = [85, 90, 95, 100, 88, 92, 96, 91]

plt.boxplot([group1, group2, group3],
            tick_labels=["Група 1", "Група 2", "Група 3"])

plt.xlabel("Групи")
plt.ylabel("Оцінки")
plt.title("Порівняння оцінок між групами")

plt.show()
```

```python
import matplotlib.pyplot as plt

dates = ["Mon", "Tue", "Wed", "Thu", "Fri", "Sat", "Sun"]
temperature = [22, 24, 23, 25, 26, 27, 24]
humidity = [60, 55, 58, 53, 50, 48, 52]

plt.plot(dates, temperature, label="Температура")
plt.plot(dates, humidity, label="Вологість")

plt.xlabel("День")
plt.ylabel("Значення")
plt.title("Температура та вологість за тиждень")
plt.legend()
plt.xticks(rotation=45)

plt.show()
```

```python
import matplotlib.pyplot as plt

hours = list(range(24))
load = [20, 18, 15, 14, 13, 16, 25, 40, 55, 60, 65, 70,
        75, 72, 68, 66, 70, 80, 85, 78, 65, 50, 35, 25]

plt.plot(hours, load)
plt.fill_between(hours, load)

plt.xlabel("Година")
plt.ylabel("Навантаження сервера")
plt.title("Погодинне навантаження сервера")
plt.grid(True)

plt.show()
```

```python
import matplotlib.pyplot as plt

days = [1, 2, 3, 4, 5, 6, 7]

conversion = [2.1, 2.3, 2.5, 2.4, 2.8, 3.0, 3.2]
retention = [40, 42, 41, 43, 45, 44, 46]
average_check = [250, 260, 255, 270, 280, 275, 290]
orders = [100, 120, 115, 130, 150, 145, 160]

fig, axs = plt.subplots(2, 2, figsize=(10, 7))

axs[0, 0].plot(days, conversion)
axs[0, 0].set_title("Конверсія")

axs[0, 1].plot(days, retention)
axs[0, 1].set_title("Утримання")

axs[1, 0].plot(days, average_check)
axs[1, 0].set_title("Середній чек")

axs[1, 1].plot(days, orders)
axs[1, 1].set_title("Кількість замовлень")

fig.suptitle("Метрики продукту")
plt.tight_layout()
 
plt.show()
```
